### 2.1. 
### Цель проекта
Цель проекта - исследовать данные о загрязнении воздуха и определить, как уровень загрязнения меняется в зависимости от территории и времени.
Для более подробного анализа будут использоваться данные по Пекину с 12 станций наблюдения. Также будут рассмотрены данные по различным городам и странам мира, чтобы сравнить показатели загрязнения между территориями.
Отдельно будет рассмотрено изменение загрязнения во времени, например по месяцам, сезонам и годам.

### Бизнес-цель
Практическая цель проекта - обработать данные о качестве воздуха и представить результаты в понятном виде, чтобы можно было увидеть, где и когда наблюдается повышенный уровень загрязнения.
В результате анализа планируется:
сравнить уровень загрязнения на разных станциях Пекина;
сравнить показатели между разными городами и странами;
определить территории с повышенным уровнем загрязнения;
сравнить показатели разных загрязняющих веществ;
посмотреть, как уровень загрязнения меняется со временем;
представить основные результаты на интерактивном дашборде.

### Границы исследования
Территориальные: разные страны и города мира; Пекин рассматривается дополнительно для более подробного анализа.
Временные: зависят от фактических периодов, представленных в выбранных источниках. Нельзя считать, что в каждом наборе есть исторические данные.
В работе будут рассматриваться такие загрязняющие вещества, как PM2.5, PM10, NO2, SO2, CO и O3, если соответствующие показатели присутствуют в выбранном источнике.
Исследование не предполагает анализ всех городов и стран мира. Результаты будут получены только для тех территорий и временных периодов, по которым имеются необходимые данные.

### Объект исследования
Объект исследования - качество атмосферного воздуха на различных территориях.
Критерии исследования
Временная закономерность - например, концентрация PM2.5 в разные сезоны.
Территориальная закономерность - например, в одних городах мира уровень загрязнения выше, чем в других. Для Пекина сравниваются данные 12 станций наблюдения; станция не считается средним значением по всему району.
Суточная закономерность - например, загрязнение повышается утром и вечером.
Связь с погодой - например, при определённой температуре или скорости ветра уровень загрязнения чаще повышается.

### Предмет исследования
Предмет исследования - изменение уровня загрязнения воздуха в зависимости от территории и времени.

### Аналитическая гипотеза
Основная гипотеза заключается в том, что уровень загрязнения воздуха имеет временные и территориальные закономерности: показатели отличаются между разными территориями и изменяются в разные периоды времени.
Для проверки территориальной части гипотезы будут сравниваться данные разных станций Пекина, а также данные по разным городам и странам.
Для проверки временной части будет рассмотрено изменение показателей загрязнения по различным временным периодам.

### В результате анализа можно будет определить, насколько заметны территориальные и временные различия в имеющихся данных.

### Функция для первичного изучения табличных данных

Для каждого CSV нужны количество строк и столбцов, первые записи, список полей и их типы, пропущенные значения, числовые характеристики и примеры категориальных полей.

```python
def overview_csv(df, label):
    print(f"НАБОР: {label}")
    print("Строк:", len(df))
    print("Столбцов:", df.shape[1])
    print("Повторяющихся строк:", int(df.duplicated().sum()))
    print("\nПервые пять строк:")
    display(df.head(5))
    print("\nСтолбцы и типы:")
    display(pd.DataFrame({"Столбец": df.columns, "Тип": [str(t) for t in df.dtypes]}))
    print("\nПропущенные значения:")
    missing = pd.DataFrame({"Пропуски": df.isna().sum(), "Доля, %": (df.isna().mean() * 100).round(2)})
    display(missing)
    numerical = df.select_dtypes(include="number")
    categorical = df.select_dtypes(include=["object", "category", "bool", "string"])
    print("\nЧисловых полей:", numerical.shape[1])
    print("Категориальных/текстовых полей:", categorical.shape[1])
    if not numerical.empty:
        print("\nОписательная статистика числовых полей:")
        display(numerical.describe().round(2))
    if categorical.shape[1]:
        print("\nПримеры значений категориальных/текстовых полей:")
        for col in list(categorical.columns)[:5]:
            print(f"{col}:", categorical[col].dropna().astype(str).unique()[:5].tolist())
```

### Источник 3. UK-AIR в формате XML

Этот набор содержит данные о качестве воздуха в Великобритании.

В нём информация хранится не в обычной таблице, а в отдельных разделах.

Этот источник нужен для изучения ещё одного формата данных.

```python
tree = ET.parse(FILES["UK-AIR XML"])
root = tree.getroot()
nodes = list(root.iter())

def local_tag(tag):
    return tag.split("}")[-1] if isinstance(tag, str) else str(tag)

print("Файл", FILES["UK-AIR XML"])
print("Главный тег", local_tag(root.tag))
print("Всего элементов", len(nodes))
print("Разных названий тегов", len(set(local_tag(e.tag) for e in nodes)))

print("\nКак выглядит начало файла в оригинале")
with open(FILES["UK-AIR XML"], "r", encoding="utf-8", errors="replace") as f:
    print(f.read(1600))

print("\nСамые частые теги")
display(pd.DataFrame(Counter(local_tag(e.tag) for e in nodes).most_common(15),
                     columns=["Тег", "Количество"]))

print("\nПримеры содержимого тегов")
examples = []
for elem in nodes:
    value = (elem.text or "").strip()
    if value:
        examples.append({"Тег": local_tag(elem.tag), "Текст": value[:120]})
    if len(examples) == 12:
        break
display(pd.DataFrame(examples))

if local_tag(root.tag).lower() == "feed":
    print("!!!ВНИМАНИЕ!!! это каталог ссылок (Atom feed), а не файл измерений.")
```

```python
numeric_xml = []
for elem in nodes:
    value = (elem.text or "").strip()
    if not value or len(value) > 60:
        continue
    try:
        number = float(value)
    except ValueError:
        continue
    numeric_xml.append({"Тег": local_tag(elem.tag), "Значение": number})

print("Отдельных числовых значений", len(numeric_xml))
if numeric_xml:
    df_xml_num = pd.DataFrame(numeric_xml)
    display(df_xml_num.head(10))
    print("Минимум и максимум по названиям тегов")
    display(df_xml_num.groupby("Тег")["Значение"].agg(["count", "min", "max"]).head(20))
else:
    print("Отдельные числовые теги не найдены.")

# В XML измерения иногда хранятся внутри одного большого текстового блока.
blocks = [e for e in nodes if local_tag(e.tag).lower() in ("values", "tuplelist") and (e.text or "").strip()]
if blocks:
    print("\nФрагмент блока измерений (пока без разделения на столбцы)")
    print((blocks[0].text or "").strip()[:500])
    print("Для этих значений понадобится отдельная обработка разделителей XML.")
else:
    print("\nБлоки values/tupleList не обнаружены.")
```

### Источник 4. OpenAQ в формате JSON

Информацию о станциях, которые следят за качеством воздуха в разных странах мира.

Сохранил в формате JSON. В файле находится информация о 100 станциях, включая их расположение и измеряемые загрязнители.

Этот источник поможет узнать, где проводятся измерения качества воздуха.

```python

with open("openaq_locations.json", "r", encoding="utf-8") as f:
    openaq = json.load(f)

locations = openaq.get("results", [])
df_openaq = pd.json_normalize(locations)

print("Разделы JSON:", list(openaq.keys()))
print("Количество станций:", len(locations))
print("Количество полей:", len(df_openaq.columns))

# Показываем основные данные без длинных списков в ячейках.
needed = [
    "id", "name", "locality", "country.name",
    "coordinates.latitude", "coordinates.longitude", "timezone"
]
available = [col for col in needed if col in df_openaq.columns]
df_stations = df_openaq[available].copy()

print("\nПервые 10 станций:")
display(df_stations.head(10))

print("\nТипы полей:")
display(df_openaq.dtypes.astype(str).to_frame("Тип данных"))

print("\nПропуски по всем полям:")
display(pd.DataFrame({
    "Пропуски": df_openaq.isna().sum(),
    "Процент, %": (df_openaq.isna().mean() * 100).round(1)
}).sort_values("Пропуски", ascending=False))
```

```python
sensor_rows = []

for station in locations:
    for sensor in station.get("sensors") or []:
        parameter = sensor.get("parameter") or {}
        if not isinstance(parameter, dict):
            parameter = {}

        name = str(parameter.get("name") or sensor.get("name") or "Неизвестно")
        simple_name = name.lower().replace(".", "").replace(" ", "")
        if simple_name == "pm25":
            name = "PM2.5"
        elif simple_name == "pm10":
            name = "PM10"

        sensor_rows.append({
            "ID станции": station.get("id"),
            "Станция": station.get("name"),
            "Показатель": name,
            "ID датчика": sensor.get("id"),
            "Единица": parameter.get("units")
        })

df_sensors = pd.DataFrame(
    sensor_rows,
    columns=["ID станции", "Станция", "Показатель", "ID датчика", "Единица"]
)

print("Количество датчиков:", len(df_sensors))
print("\nПервые 10 датчиков:")
display(df_sensors.head(10))

if not df_sensors.empty:
    print("\nКоличество датчиков по показателям:")
    display(df_sensors["Показатель"].value_counts().to_frame("Количество"))

    # PM10 и PM2.5 — отдельные столбцы.
    # Если у станции несколько датчиков одного типа, показываем их все.
    pm = df_sensors[df_sensors["Показатель"].isin(["PM10", "PM2.5"])].copy()
    if not pm.empty:
        pm["ID датчика"] = pm["ID датчика"].astype(str)
        pm_wide = (pm.groupby(["ID станции", "Станция", "Показатель"], dropna=False)["ID датчика"]
                     .agg(", ".join)
                     .unstack("Показатель")
                     .reset_index())
        for column in ["PM10", "PM2.5"]:
            if column not in pm_wide.columns:
                pm_wide[column] = None
        pm_wide = pm_wide[["ID станции", "Станция", "PM10", "PM2.5"]]
        print("\nPM10 и PM2.5 по отдельным столбцам:")
        display(pm_wide.head(15))
    else:
        print("Датчиков PM10 и PM2.5 в этой выгрузке нет.")
else:
    print("Датчики в файле не найдены.")

print("\nЧисла PM10 и PM2.5 в таблице — ID датчиков, а не значения загрязнения.")
```

```python
numeric = df_openaq.select_dtypes(include="number").drop(columns=["id", "owner.id", "provider.id"], errors="ignore").dropna(axis=1, how="all")

print("Числовые поля (без служебных ID):")
print(numeric.columns.tolist())

if not numeric.empty:
    print("\nМинимум, максимум и среднее:")
    display(numeric.agg(["min", "max", "mean"]).round(2))

print("\nСколько станций имеют датчики PM10 и PM2.5:")
if not df_sensors.empty:
    coverage = (df_sensors[df_sensors["Показатель"].isin(["PM10", "PM2.5"])]
                .groupby("Показатель")["ID станции"].nunique())
    display(coverage.reindex(["PM10", "PM2.5"], fill_value=0).to_frame("Количество станций"))
else:
    print("Нет сведений о датчиках.")
```

### Источник 5. Текст WHO из HTML

Статья о загрязнении воздуха и его влиянии на здоровье человека.

Информация представлена в основном в виде обычного текста.

```python
with open(FILES["WHO HTML"], "r", encoding="utf-8", errors="replace") as f:
    soup = BeautifulSoup(f.read(), "html.parser")
page_title = soup.title.get_text(" ", strip=True) if soup.title else "Заголовок не найден"
for unwanted in soup(["script", "style", "nav", "footer", "header", "noscript"]):
    unwanted.decompose()
main = soup.find("main") or soup.find("article") or soup.body or soup
who_text = main.get_text(" ", strip=True)
words = re.findall(r"[A-Za-zА-Яа-яЁё]+", who_text.lower())
stopwords = {"the", "and", "for", "that", "with", "from", "this", "are", "was", "have", "has", "not", "who", "can", "also", "air", "its", "all", "but", "their", "more", "than", "into", "such", "these", "will", "which", "were", "about", "each", "other", "between", "some"}
useful_words = [w for w in words if len(w) >= 4 and w not in stopwords]
print("Заголовок", page_title)
print("Всего слов", len(words))
print("Уникальных слов", len(set(words)))
print("\nНачало текста")
print(who_text[:1200])
print("\nЧасто встречающиеся слова (без части служебных слов)")
display(pd.DataFrame(Counter(useful_words).most_common(15), columns=["Слово", "Частота"]))
if len(words) < 100:
    print("не то.")
```

### 1. Проверка качества и очистка CSV

Сначала смотрю размеры, примеры, типы, пропуски и повторы. Названия стран и городов нужны для сравнения территорий, координаты - для карты. AQI - индекс качества воздуха, а не концентрация. В Пекине дата нужна для временных сравнений, станция - для сравнения мест, погода - для поиска связей. В файле Пекина year, month, day и hour объединяются в datetime.
Убираю только полные повторы, лишние пробелы и пустые строки. Ошибочные числа и координаты отмечаю как пропуски. Отрицательные значения индексов и концентраций считаю ошибками Очень большие значения автоматически не удаляю тк это могут быть реальные эпизоды. Пропуски не заменяю нулями или средними, чтобы не придумывать наблюдения. Нераспознанные поля остаются в очищенной таблице.


```python
def key23(name):
    return re.sub(r'[^a-z0-9]', '', str(name).lower())

def duplicate_view23(df):
    result = df.copy()
    for col in result.columns:
        if result[col].dtype == 'object' or pd.api.types.is_string_dtype(result[col]):
            result[col] = result[col].map(
                lambda value: json.dumps(value, sort_keys=True, ensure_ascii=False)
                if isinstance(value, (list, dict)) else value)
    return result

def quality23(df, label):
    print(label, '- строк', len(df), 'столбцов', len(df.columns),
          'полных повторо', int(duplicate_view23(df).duplicated().sum()))
    display(df.head())
    display(pd.DataFrame({'Тип': df.dtypes.astype(str),
                          'Пропуски': df.isna().sum(),
                          'Пропуски, %': (100 * df.isna().mean()).round(2)}))
    numbers = df.select_dtypes(include='number')
    if len(numbers.columns):
        display(numbers.describe().T)

pollutants23 = {'pm25', 'pm1', 'pm10', 'no2', 'so2', 'co', 'o3', 'ozone', 'no', 'nox'}
clean23, audit23, dates23 = {}, [], {}
for source, raw in raw23.items():
    quality23(raw, source + ': до очистки')
    df = raw.copy(deep=True)
    for col in df.select_dtypes(include=['object', 'string']).columns:
        df[col] = df[col].astype('string').str.strip().replace('', pd.NA)
    for col in df.columns:
        k = key23(col)
        is_measure = k in pollutants23 or k.endswith('aqivalue')
        limits = (-90, 90) if k in ('lat', 'latitude') else ((-180, 180) if k in ('lng', 'lon', 'longitude') else None)
        if is_measure or limits or k in ('humidity', 'relativehumidity', 'temperature', 'temp', 'pres', 'dewp', 'rain', 'wspm'):
            old = df[col].copy()
            num = pd.to_numeric(old, errors='coerce')
            invalid = old.notna() & num.isna()
            invalid |= num.isin([float('inf'), -float('inf')])
            if is_measure:
                invalid |= num.lt(0)
            if limits:
                invalid |= ~num.between(*limits) & num.notna()
            if k in ('rain', 'wspm', 'pres'):
                invalid |= num.lt(0)
            if k in ('humidity', 'relativehumidity'):
                invalid |= ~num.between(0, 100) & num.notna()
            audit23.append({'source': source, 'field': col, 'action': 'числовые ошибки в пропуски', 'count': int(invalid.sum())})
            df[col] = num.mask(invalid)
        if k in ('date', 'datetime', 'timestamp', 'time', 'createdat'):
            if not pd.api.types.is_numeric_dtype(df[col]):
                parsed = pd.to_datetime(df[col], errors='coerce', format='mixed')
                dates23[(source, col)] = parsed
                audit23.append({'source': source, 'field': col, 'action': 'не распознана дата; оригинал сохранён', 'count': int((df[col].notna() & parsed.isna()).sum())})
    duplicate_count = int(duplicate_view23(df).duplicated().sum())
    df = df.loc[~duplicate_view23(df).duplicated()].copy()
    audit23.append({'source': source, 'field': '*', 'action': 'удалены полные повторы', 'count': duplicate_count})
    clean23[source] = df
    quality23(df, source + ' после очистки')
audit23 = pd.DataFrame(audit23)
display(audit23)

beijing_pm25_23 = pd.DataFrame()
beijing_weather_23 = pd.DataFrame()
beijing_ready23 = clean23['beijing']
if 'PM2.5' in beijing_ready23.columns:
    columns = [c for c in ['datetime', 'station', 'PM2.5']
               if c in beijing_ready23.columns]
    beijing_pm25_23 = beijing_ready23[columns].dropna(subset=['PM2.5']).copy()
    print('Пекин: строк с известным PM2.5', len(beijing_pm25_23))
    print('Строк без PM2.5, не включённых в эту выборку',
          int(beijing_ready23['PM2.5'].isna().sum()))
    display(beijing_pm25_23.head())
    weather = [c for c in ['TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM'] if c in beijing_ready23]
    if weather:
        columns += [c for c in weather if c not in columns]
        beijing_weather_23 = beijing_ready23[columns].dropna(subset=['PM2.5'] + weather).copy()
        print('Строк с PM2.5 и всеми выбранными погодными полями', len(beijing_weather_23))
        print('Это отдельная неполная выборка: она может охватывать другие станции и периоды.')
```

### 2. JSON: станции и датчики

Разделяю вложенный JSON на две таблицы. Одна строка в stations - станция, одна строка в sensors - датчик.= Сохраняю все поля через нормализацию, а не только PM10 и PM2.5. Также проверяю, есть ли датчики без станции и повторяющиеся айди. Если выгрузка содержит только часть станций, выводы относятся только к этой части.

```python
with paths23['json'].open(encoding='utf-8') as f:
    payload23 = json.load(f)
locations23 = payload23.get('results', []) if isinstance(payload23, dict) else payload23
if not isinstance(locations23, list):
    raise ValueError('В JSON ожидается список станций или results со списком')

def split_locations23(locations):
    groups = {}
    def flatten(record, prefix, station_row, station_id, row_path):
        flat = {}
        for name, value in record.items():
            field = prefix + '.' + name if prefix else name
            if isinstance(value, dict):
                flat.update(flatten(value, field, station_row, station_id, row_path))
            elif isinstance(value, list):
                for position, item in enumerate(value):
                    item_path = row_path + '/' + field + '/' + str(position)
                    row = {'station_row': station_row, 'station_id': station_id,
                           'parent_path': row_path, 'row_path': item_path, 'position': position}
                    if isinstance(item, dict):
                        row.update(flatten(item, '', station_row, station_id, item_path))
                    elif isinstance(item, list):
                        row.update(flatten({'items': item}, field, station_row, station_id, item_path))
                    else:
                        row['value'] = item
                    groups.setdefault(field, []).append(row)
            else:
                flat[field] = value
        return flat
    station_rows = []
    for index, station in enumerate(locations):
        row = flatten(station, '', index, station.get('id'), 'station/' + str(index))
        row['station_row'] = index
        station_rows.append(row)
    stations = pd.DataFrame(station_rows)
    details = {}
    for index, (field, rows) in enumerate(sorted(groups.items())):
        name = 'openaq_detail_' + str(index) + '_' + re.sub(r'[^a-zA-Z0-9_]', '_', field)
        details[name] = pd.DataFrame(rows)
    return stations, details

stations23, openaq_details23 = split_locations23(locations23)
sensor_records23 = []
for station in locations23:
    for sensor in station.get('sensors') or []:
        sensor_records23.append({'station_id': station.get('id'), 'sensor': sensor})
sensors23 = pd.json_normalize(sensor_records23)
if sensors23.empty:
    sensors23 = pd.DataFrame(columns=['station_id', 'sensor.id'])
quality23(stations23, 'Станции OpenAQ')
print('Датчико', len(sensors23))
display(sensors23.head(10))
for detail_name, detail in openaq_details23.items():
    if detail_name.endswith('_sensors'):
        continue  # Таблица датчиков выше
    print('Отдельная таблица', detail_name, '— строк:', len(detail))
    display(detail.head(10))
if 'id' in stations23:
    print('Повторяющихся ID станций', int(stations23['id'].dropna().duplicated().sum()))
    print('Датчиков без известной станции', int((~sensors23['station_id'].isin(stations23['id'])).sum()))
if 'sensor.id' in sensors23:
    print('Повторяющихся ID датчиков', int(sensors23['sensor.id'].dropna().duplicated().sum()))
```

### 3. XML: структура, значения и ограничения

Не считаю каждый числовой тег концентрацией: это может быть код или номер. Сохраняю путь, атрибуты и текст каждого элемента, чтобы не потерять контекст. Отдельно разбираю SWE DataArray по описанию полей и разделителей самого XML. Если описание неполное или ширина строки не совпадает, сохраняю исходный блок и причину, а не назначаю столбцы наугад. Такие строки пока нельзя объединять с измерениями CSV. Atom feed рассматриваю как каталог ссылок.

```python
root23 = ET.parse(paths23['xml']).getroot()
def tag23(e):
    return e.tag.split('}')[-1] if isinstance(e.tag, str) else ''
xml_rows23 = []
def walk23(e, path=''):
    here = path + '/' + tag23(e)
    xml_rows23.append({'path': here, 'tag': tag23(e),
                       'attributes': json.dumps(e.attrib, ensure_ascii=False),
                       'text': (e.text or '').strip()})
    for child in e:
        walk23(child, here)
walk23(root23)
xml_nodes23 = pd.DataFrame(xml_rows23)
display(xml_nodes23.head(15))
xml_values23, xml_issues23 = [], []
for block_id, array in enumerate(e for e in root23.iter() if tag23(e) == 'DataArray'):
    fields = [e for e in array.iter() if tag23(e) == 'field']
    enc = next((e for e in array.iter() if tag23(e) == 'TextEncoding'), None)
    value = next((e for e in array.iter() if tag23(e) == 'values'), None)
    if enc is None or value is None or not fields:
        xml_issues23.append({'block': block_id, 'reason': 'Нет полей, TextEncoding или values'})
        continue
    token, block = enc.get('tokenSeparator'), enc.get('blockSeparator')
    decimal = enc.get('decimalSeparator', '.')
    if not token or not block:
        xml_issues23.append({'block': block_id, 'reason': 'Нет явных разделителей'})
        continue
    names = [f.get('name', 'field_' + str(i)) for i, f in enumerate(fields)]
    for row_id, row in enumerate((value.text or '').strip().split(block)):
        if not row.strip():
            continue
        parts = row.strip().split(token)
        if len(parts) != len(fields):
            xml_issues23.append({'block': block_id, 'row': row_id, 'reason': 'Число значений не совпало с числом полей', 'raw': row})
            continue
        for field_id, (field, name, text) in enumerate(zip(fields, names, parts)):
            quantity = next((e for e in field.iter() if tag23(e) == 'Quantity'), None)
            unit = next((e.get('code') or e.get('href') for e in field.iter() if tag23(e) == 'uom'), None)
            nils = {(e.text or '').strip() for e in field.iter() if tag23(e) == 'nilValue'}
            number = None
            if quantity is not None and text.strip() not in nils:
                try:
                    number = float(text.strip().replace(decimal, '.'))
                    if not float('-inf') < number < float('inf'):
                        number = None
                except ValueError:
                    pass
            xml_values23.append({'block': block_id, 'row': row_id, 'field_id': field_id,
                                 'field': name, 'raw': text, 'numeric': number, 'unit': unit,
                                 'definition': quantity.get('definition') if quantity is not None else None})
xml_values23 = pd.DataFrame(xml_values23, columns=['block','row','field_id','field','raw','numeric','unit','definition'])
xml_issues23 = pd.DataFrame(xml_issues23, columns=['block','row','reason','raw'])
print('Разобрано значений SWE:', len(xml_values23))
display(xml_values23.head(15))
display(xml_issues23)
if tag23(root23).lower() == 'feed':
    print('Это каталог Atom, измерений в нём может не быть.')
if xml_values23.empty:
    print('Поддерживаемые SWE-блоки не найдены. Структура и исходные значения сохранены в xml_nodes; числовые коды не выданы за измерения.')
```

### 4. HTML, разметка текста и векторизация

Из статьи WHO убираю служебные элементы страницы. Каждый абзац - отдельный документ. По словам отмечаю темы: здоровье, загрязнители и меры снижения загрязнения. TF-IDF превращает абзацы в числовые признаки, а высокий вес означает, что слово характерно для абзаца. Текст помогает объяснять тему.

```python
soup23 = BeautifulSoup(paths23['html'].read_text(encoding='utf-8'), 'html.parser')
for e in soup23(['script', 'style', 'nav', 'footer', 'header', 'noscript']):
    e.decompose()
main23 = soup23.find('article') or soup23.find('main') or soup23.body or soup23
rules23 = {
    'здоровье': ['health', 'disease', 'death', 'cancer', 'stroke', 'respiratory'],
    'загрязнители': ['pm2.5', 'pm10', 'ozone', 'nitrogen dioxide', 'particulate', 'carbon monoxide'],
    'меры': ['transport', 'energy', 'policy', 'policies', 'emission', 'fuel']
}
paragraphs23 = []
for p in main23.find_all('p'):
    text = p.get_text(' ', strip=True)
    if not text:
        continue
    matches = {topic: [w for w in terms if re.search(r'(?<!\w)' + re.escape(w) + r'(?!\w)', text.lower())]
               for topic, terms in rules23.items()}
    paragraphs23.append({'paragraph_id': len(paragraphs23), 'text': text,
                          'topics': '; '.join(t for t, terms in matches.items() if terms) or 'прочее',
                          'evidence': json.dumps(matches, ensure_ascii=False)})
text23 = pd.DataFrame(paragraphs23, columns=['paragraph_id','text','topics','evidence'])
display(text23.head(10))
tfidf23 = pd.DataFrame(columns=['paragraph_id', 'term', 'weight'])
if len(text23) and text23['text'].str.contains(r'[A-Za-zА-Яа-я]{2,}', regex=True).any():
    vectorizer23 = TfidfVectorizer(stop_words='english', max_features=1000)
    try:
        matrix23 = vectorizer23.fit_transform(text23['text'])
        coo23 = matrix23.tocoo()
        terms23 = vectorizer23.get_feature_names_out()
        tfidf23 = pd.DataFrame({'paragraph_id': coo23.row, 'term': terms23[coo23.col], 'weight': coo23.data})
        print('Размер матрицы TF-IDF:', matrix23.shape)
        display(tfidf23.sort_values('weight', ascending=False).head(15))
    except ValueError as error:
        print('Векторизация не выполнена:', error)
else:
    print('Нет подходящих абзацев')
```

### 5. Общая структура, объединение и обновление

Сохраняю по одному подготовленному файлу на источник: два CSV и три JSON. CSV удобен для обычных таблиц, а JSON позволяет сложить несколько связанных таблиц в один файл. Отдельная база для этой работы не нужна.


```python

BEIJING_UNITS23 = {c: 'µg/m³' for c in ['PM2.5','PM10','SO2','NO2','CO','O3']}
AVERAGING23 = {'beijing': '1 hour'}
aliases23 = {'pm25':'PM2.5', 'pm1':'PM1', 'pm10':'PM10', 'no2':'NO2',
             'so2':'SO2', 'co':'CO', 'o3':'O3', 'ozone':'O3', 'no':'NO', 'nox':'NOx'}
parts23, mapping_rows23 = [], []
for source, df in clean23.items():
    fields = {key23(c): c for c in df.columns}
    def find23(*names):
        return next((fields[k] for k in names if k in fields), None)
    country, city = find23('country', 'countryname'), find23('city')
    station = find23('station', 'stationid', 'sensorid', 'locationid', 'sitename')
    date = find23('datetime', 'timestamp', 'date', 'time', 'createdat')
    for col in df.columns:
        k = key23(col)
        metric, kind, unit = None, None, None
        if k.endswith('aqivalue') or k == 'aqi':
            metric, kind, unit = str(col), 'index', 'AQI'
        elif k in aliases23 and not (source == 'beijing' and col == 'No'):
            metric, kind = aliases23[k], 'concentration'
            unit = BEIJING_UNITS23.get(col) if source == 'beijing' else None
        mapping_rows23.append({'source':source,'field':col,'metric':metric,
                              'role':kind or 'контекст или непроверенное поле','unit':unit})
        if metric is None:
            continue
        values = pd.to_numeric(df[col], errors='coerce')
        valid = values.notna() & values.ge(0) & values.lt(float('inf'))
        rows = df.loc[valid]
        part = pd.DataFrame(index=rows.index)
        part['source'] = source
        part['source_row'] = rows.index
        part['country'] = rows[country] if country else ('China' if source == 'beijing' else None)
        part['city'] = rows[city] if city else ('Beijing' if source == 'beijing' else None)
        part['station'] = rows[station] if station else None
        part['time_raw'] = rows[date].astype('string') if date else None
        part['metric'], part['kind'] = metric, kind
        part['value'], part['unit'] = values.loc[valid].astype(float), unit
        part['averaging_period'] = AVERAGING23.get(source)
        parts23.append(part)
observations23 = pd.concat(parts23,ignore_index=True) if parts23 else pd.DataFrame(
    columns=['source','source_row','country','city','station','time_raw','metric','kind','value','unit','averaging_period'])
mapping23 = pd.DataFrame(mapping_rows23)
display(mapping23)
print('Объединённая длинная таблица:', len(observations23), 'значений')
display(observations23.head(10))
display(observations23.groupby(['source','metric','kind'],dropna=False).size().to_frame('count'))


def records23(df):
    return json.loads(df.to_json(orient='records', force_ascii=False, date_format='iso', double_precision=15))

def save_source23(path, data):
    if isinstance(data, pd.DataFrame):
        data.to_csv(path, index=False, encoding='utf-8-sig')
    else:
        path.write_text(json.dumps(data, ensure_ascii=False, separators=(',', ':'), allow_nan=False), encoding='utf-8')

prepared23 = {
    'global_clean.csv': clean23['global'],
    'beijing_clean.csv': clean23['beijing'],
    'openaq_prepared.json': {
        'stations': records23(stations23),
        'details': {name: records23(frame) for name, frame in openaq_details23.items()}
    },
    'uk_air_prepared.json': {
        'structure': records23(xml_nodes23),
        'values': records23(xml_values23),
        'issues': records23(xml_issues23)
    },
    'who_prepared.json': {
        'paragraphs': records23(text23),
        'tfidf': records23(tfidf23)
    }
}

file_report23 = []
for filename, data in prepared23.items():
    path = OUT / filename
    save_source23(path, data)
    file_report23.append({'Файл': filename, 'Размер, МБ': round(path.stat().st_size / 1024**2, 2)})
    if isinstance(data, pd.DataFrame):
        assert len(pd.read_csv(path, low_memory=False)) == len(data)
    else:
        assert json.loads(path.read_text(encoding='utf-8')) == data

print('Сохранены пять подготовленных файлов:')
display(pd.DataFrame(file_report23))
print('Журнал очистки CSV:')
display(audit23)
for label, frame in [
    ('Глобальные данные', clean23['global']), ('Пекин', clean23['beijing']),
    ('Станции OpenAQ', stations23), ('Датчики OpenAQ', sensors23),
    ('Структура XML', xml_nodes23), ('Значения XML', xml_values23),
    ('Текст WHO', text23), ('Признаки WHO', tfidf23)
]:
    print(label, '— строк:', len(frame))
    display(frame.head(5))
assert not observations23.duplicated(['source','source_row','metric']).any()
print('Общая таблица в памяти: observations23, значений:', len(observations23))
print('Результаты:', OUT.resolve())
```
